```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from useful_func import *
from sklearn.preprocessing import LabelEncoder , StandardScaler
```

```python
df = pd.read_csv("customer_segmentation_uncleaned_data.csv")
df.head()
```

```python
# fill missing vaalues
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Annual_Income_k$"] = df["Annual_Income_k$"].fillna(df["Annual_Income_k$"].mean())
```

```python
# fill outliers
```

```python
df["Annual_Income_k$"] = df["Annual_Income_k$"].clip(lower=0, upper=200)
```

```python
# # Encoding : Gender columns to convert label Encoder
```

```python
le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])
```

```python
# Scaling : 
```

```python
scaler = StandardScaler()
cols_to_scale = ["Age", "Annual_Income_k$", "Spending_Score", "Purchase_Frequency"]
df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])
```

```python
# #import csv
df.to_csv("Cleaned_Data.csv", index=False)
```
